```python
def word_count(sentence):
    """How many words are in the sentence."""
    return len(sentence.split())

def longest_word(sentence):
    """The longest word in the sentence."""
    longest = ""
    for w in sentence.split():
        if len(w) > len(longest):
            longest = w
    return longest

def count_vowels(text):
    """How many vowels (a e i o u) are in the text."""
    vowels = "aeiou"
    total = 0
    for letter in text.lower():
        if letter in vowels:
            total += 1
    return total

def long_words(sentence, min_len=4):
    """Every word longer than min_len, in CAPITALS."""
    return [w.upper() for w in sentence.split() if len(w) > min_len]

def initials(full_name):
    """Turn 'erbert suratos limbos' into 'E.S.L.'"""
    return ".".join([part[0].upper() for part in full_name.split()]) + "."

s = "the quick brown fox jumps"
print(word_count(s), longest_word(s), count_vowels(s))
print(long_words(s))
print(initials("erbert suratos limbos"))
```
